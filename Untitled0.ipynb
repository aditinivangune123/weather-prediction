```python
import re
import spacy
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

print("spaCy model loaded successfully!")
```

```python
def preprocess_text(text):
    """
    Preprocess input text using spaCy.
    - Converts text to lowercase
    - Removes stop words
    - Removes punctuation
    - Removes numbers
    - Lemmatizes words
    """

    doc = nlp(text)

    cleaned_words = []

    for token in doc:
        if (
            not token.is_stop
            and not token.is_punct
            and not token.is_space
            and not token.like_num
        ):
            lemma = token.lemma_.lower().strip()

            if lemma and len(lemma) > 1:
                cleaned_words.append(lemma)

    return " ".join(cleaned_words)
```

```python
def get_sentences(text):
    """
    Split the document into sentences using spaCy.
    """

    doc = nlp(text)

    sentences = [
        sent.text.strip()
        for sent in doc.sents
        if len(sent.text.strip()) > 20
    ]

    return sentences
```

```python
def calculate_tfidf(sentences):
    """
    Convert sentences into TF-IDF vectors.
    """

    processed_sentences = [
        preprocess_text(sentence)
        for sentence in sentences
    ]

    vectorizer = TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 2)
    )

    tfidf_matrix = vectorizer.fit_transform(processed_sentences)

    return tfidf_matrix, vectorizer
```

```python
def calculate_sentence_scores(sentences, tfidf_matrix):
    """
    Calculate importance score for every sentence.
    """

    # TF-IDF importance
    tfidf_scores = np.asarray(
        tfidf_matrix.sum(axis=1)
    ).flatten()

    # Normalize TF-IDF scores
    if tfidf_scores.max() > 0:
        tfidf_scores = (
            tfidf_scores / tfidf_scores.max()
        )

    # Position score
    position_scores = []

    total_sentences = len(sentences)

    for i in range(total_sentences):

        # Earlier sentences get slightly higher importance
        position_score = 1 - (i / total_sentences)

        position_scores.append(position_score)

    position_scores = np.array(position_scores)

    # Sentence length score
    length_scores = []

    for sentence in sentences:

        word_count = len(sentence.split())

        # Prefer sentences with reasonable length
        if word_count < 5:
            score = 0.3

        elif word_count > 50:
            score = 0.6

        else:
            score = 1.0

        length_scores.append(score)

    length_scores = np.array(length_scores)

    # Final weighted score
    final_scores = (
        0.60 * tfidf_scores +
        0.25 * position_scores +
        0.15 * length_scores
    )

    return final_scores
```

```python
def remove_redundancy(
    selected_sentences,
    selected_vectors,
    threshold=0.75
):
    """
    Remove highly similar/redundant sentences.
    """

    final_sentences = []
    final_vectors = []

    for sentence, vector in zip(
        selected_sentences,
        selected_vectors
    ):

        is_redundant = False

        for previous_vector in final_vectors:

            similarity = cosine_similarity(
                vector,
                previous_vector
            )[0][0]

            if similarity >= threshold:
                is_redundant = True
                break

        if not is_redundant:
            final_sentences.append(sentence)
            final_vectors.append(vector)

    return final_sentences
```

```python
def summarize_text(text, summary_ratio=0.30):
    """
    Complete NLP extractive summarization pipeline.
    """

    # -----------------------------
    # INPUT VALIDATION
    # -----------------------------

    if text is None:
        return "Please enter some text."

    text = text.strip()

    if len(text) < 100:
        return "Please enter a longer text (at least 100 characters)."

    # -----------------------------
    # SENTENCE SEGMENTATION
    # -----------------------------

    sentences = get_sentences(text)

    if len(sentences) < 3:
        return "Please enter text containing at least 3 sentences."

    # -----------------------------
    # FEATURE EXTRACTION
    # -----------------------------

    tfidf_matrix, vectorizer = calculate_tfidf(sentences)

    # -----------------------------
    # SENTENCE SCORING
    # -----------------------------

    scores = calculate_sentence_scores(
        sentences,
        tfidf_matrix
    )

    # -----------------------------
    # NUMBER OF SENTENCES
    # -----------------------------

    number_of_sentences = max(
        2,
        int(len(sentences) * summary_ratio)
    )

    number_of_sentences = min(
        number_of_sentences,
        len(sentences)
    )

    # -----------------------------
    # SELECT TOP SENTENCES
    # -----------------------------

    ranked_indices = np.argsort(
        scores
    )[::-1]

    selected_indices = ranked_indices[
        :number_of_sentences
    ]

    # -----------------------------
    # SORT ACCORDING TO ORIGINAL
    # ORDER
    # -----------------------------

    selected_indices = sorted(
        selected_indices
    )

    selected_sentences = [
        sentences[i]
        for i in selected_indices
    ]

    selected_vectors = [
        tfidf_matrix[i]
        for i in selected_indices
    ]

    # -----------------------------
    # REMOVE REDUNDANCY
    # -----------------------------

    final_sentences = remove_redundancy(
        selected_sentences,
        selected_vectors
    )

    # -----------------------------
    # OUTPUT
    # -----------------------------

    summary = " ".join(final_sentences)

    return summary
```

```python
sample_text = """
Artificial intelligence is transforming many industries around the world.
Natural language processing is a branch of artificial intelligence that enables
computers to understand and process human language. Text summarization is an
important NLP application because large amounts of information are generated
every day. Reading long documents manually can consume a significant amount
of time. An automatic summarization system can identify important information
and generate a shorter version of the original document. Extractive
summarization selects important sentences directly from the original text.
TF-IDF can be used to identify important words in a document. Sentences
containing important words can receive higher scores during summarization.
spaCy can be used for tokenization, sentence segmentation, stop-word removal
and linguistic preprocessing. Scikit-learn provides machine learning tools
such as TF-IDF vectorization and cosine similarity. Combining these NLP
techniques allows us to create an efficient text summarization system.
Such a system can be useful for news articles, research papers, reports,
emails and other large documents.
"""

summary = summarize_text(sample_text)

print("ORIGINAL TEXT:\n")
print(sample_text)

print("\n" + "="*80)

print("\nGENERATED SUMMARY:\n")
print(summary)
```

```python
import gradio as gr
```

```python
def real_time_summarizer(text, ratio):

    if not text or len(text.strip()) < 100:
        return "Enter a longer text to generate a summary."

    try:
        summary = summarize_text(
            text,
            summary_ratio=ratio
        )

        return summary

    except Exception as e:
        return f"Error: {str(e)}"
```

```python
demo = gr.Interface(
    fn=real_time_summarizer,

    inputs=[
        gr.Textbox(
            lines=15,
            placeholder="Paste your article, report, research paper, news article, etc.",
            label="Input Text"
        ),

        gr.Slider(
            minimum=0.10,
            maximum=0.60,
            value=0.30,
            step=0.05,
            label="Summary Ratio"
        )
    ],

    outputs=gr.Textbox(
        lines=10,
        label="Generated Summary"
    ),

    title="📝 Real-Time NLP Text Summarization System",

    description=(
        "Extractive text summarization using "
        "spaCy + TF-IDF + scikit-learn + NLP"
    ),

    live=True
)

demo.launch(
    share=True
)
```
